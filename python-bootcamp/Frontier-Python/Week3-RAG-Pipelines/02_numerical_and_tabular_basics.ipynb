## 1. `pandas` fundamentals

```python
import pandas as pd

claims = pd.DataFrame({
    "claim_id": ["C-1001", "C-1002", "C-1003", "C-1004"],
    "policy_id": ["P-55", "P-12", "P-55", "P-90"],
    "status": ["approved", "rejected", "approved", "approved"],
    "amount": [15000, 0, 8000, 22000],
})

print(claims)
print()

# Filtering
approved = claims[claims["status"] == "approved"]
print("Approved claims:\n", approved)

# Grouping — e.g. total approved amount per policy
totals = claims[claims.status == "approved"].groupby("policy_id")["amount"].sum()
print("\nTotals by policy:\n", totals)
```

## 2. `numpy` fundamentals — and why it matters for embeddings

```python
import numpy as np

# An "embedding" is just a vector of numbers. Real ones have hundreds of dimensions;
# we use 4 here so the math is easy to see.
doc_vectors = {
    "motor_policy_clause_4_2": np.array([0.10, 0.85, 0.05, 0.20]),
    "irdai_circular_timelines": np.array([0.75, 0.05, 0.60, 0.10]),
    "health_policy_exclusions": np.array([0.15, 0.80, 0.10, 0.25]),
}

query_vector = np.array([0.12, 0.83, 0.08, 0.22])   # "what does the exclusion clause say?"

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

scores = {name: cosine_similarity(query_vector, vec) for name, vec in doc_vectors.items()}
ranked = sorted(scores.items(), key=lambda kv: kv[1], reverse=True)

print("Ranked retrieval results:")
for name, score in ranked:
    print(f"  {score:.4f}  {name}")
```

This *is* the core operation every vector database (Chroma, Qdrant, pgvector) performs —
they just do it over millions of vectors, with an index instead of a linear scan.
Understanding this loop is what makes "top-k retrieval" a concrete idea instead of a black box.

## 3. List and dict comprehensions

```python
raw_chunks = [
    "  4.1 Wear and tear is excluded.  ",
    "",
    "4.2 Pre-existing damage is excluded.",
    "   ",
]

# List comprehension: clean + drop empties, in one line
clean_chunks = [c.strip() for c in raw_chunks if c.strip()]
print(clean_chunks)

# Dict comprehension: tag each chunk with its length, for chunk-size auditing
chunk_lengths = {c: len(c) for c in clean_chunks}
print(chunk_lengths)
```
