## 1. Variables, types, and coercion

```python
age = 5          # int
rate = 3.5       # float
name = "Nexus"   # str
active = True    # bool

# Coercion — this is a common early bug source
print(str(age) + " years")      # must coerce int -> str to concatenate
print(age + int("10"))          # must coerce str -> int to add
print(type(age), type(rate), type(name), type(active))
```

## 2. Control flow

```python
scores = [92, 55, 78, 40, 88]

for s in scores:
    if s >= 80:
        label = "high"
    elif s >= 60:
        label = "medium"
    else:
        label = "low"
    print(s, "->", label)

# while loop with a break condition
total, i = 0, 0
while True:
    total += scores[i]
    i += 1
    if i == len(scores):
        break
print("sum:", total)
```

## 3. Functions — defaults, *args, **kwargs

```python
# *args → collects positional arguments into a tuple
def add(*numbers):
    print(type(numbers))
    return sum(numbers)

print(add(10, 20, 30))
# numbers = (10, 20, 30)


# **kwargs → collects keyword arguments into a dictionary
def show(**details):
    print(type(details))
    print(details)

show(name="Naresh", age=49)
# details = {"name": "Naresh", "age": 49}
```

```python
def build_prompt(task, tone="neutral", **extra_fields):
    lines = [f"Task: {task}", f"Tone: {tone}"]
    print(type(lines))
    print(extra_fields)
    for key, value in extra_fields.items():
        lines.append(f"{key}: {value}")
    return "\n".join(lines)

#print(build_prompt("Summarize the ticket"))
print()
print(build_prompt("Summarize the ticket", tone="formal", audience="client", max_words=50))

def total_tokens(*counts):
    return sum(counts)

print("\ntotal_tokens:", total_tokens(120, 340, 15))
```

## 4. Core data structures

```python
# list — ordered, mutable
fields = ["query_type", "urgency", "submitting_team"]
fields.append("regulation_ref")
print("list:", fields)

# tuple — ordered, immutable (good for fixed records)
record = ("SmartIntake", "Week2", "PharmaCore India")
print("tuple:", record)

# dict — key/value, the shape most LLM API responses come back as
extracted = {"query_type": "labelling", "urgency": "high"}
extracted["jurisdiction"] = "India"   # add a key
print("dict:", extracted)

# set — unique values, fast membership checks
seen_orgs = {"PharmaCore India", "FinBridge Solutions"}
print("is NeoBank seen?", "NeoBank India" in seen_orgs)
```

## 5. String formatting (f-strings)

```python
org = "PharmaCore India"
urgency = "high"
score = 0.873

# f-strings are the standard for prompt templating
print(f"Org: {org} | Urgency: {urgency.upper()} | Confidence: {score:.2%}")
```
