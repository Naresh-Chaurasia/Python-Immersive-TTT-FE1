## 2. Basic OOP — classes and inheritance

````python
class IntakeRequest:
    def __init__(self, org, query_type):
        self.org = org
        self.query_type = query_type   # instance attribute

    def summary(self):
        return f"{self.org}: {self.query_type}"

class UrgentIntakeRequest(IntakeRequest):   # inheritance
    def __init__(self, org, query_type, escalate_to):
        super().__init__(org, query_type)  # reuse the parent constructor
        self.escalate_to = escalate_to

    def summary(self):   # override
        base = super().summary()
        return f"[URGENT -> {self.escalate_to}] {base}"

r1 = IntakeRequest("PharmaCore India", "labelling")
r2 = UrgentIntakeRequest("FinBridge Solutions", "SAR filing", "compliance-lead")

for r in (r1, r2):
    print(r.summary())
````

## 3. Exception handling

````python
def parse_urgency(value):
    valid = {"low", "medium", "high"}
    if value not in valid:
        raise ValueError(f"'{value}' is not a valid urgency level")
    return value

for candidate in ["high", "urgent", "low"]:
    try:
        print(candidate, "->", parse_urgency(candidate))
    except ValueError as e:
        print(candidate, "-> REJECTED:", e)
    finally:
        pass   # place for cleanup code, if any
````

## 4. File I/O

````python
import tempfile, os

tmp_dir = tempfile.mkdtemp()
log_path = os.path.join(tmp_dir, "intake_log.txt")

with open(log_path, "w") as f:
    f.write("PharmaCore India | labelling | high\n")
    f.write("FinBridge Solutions | SAR filing | high\n")

with open(log_path, "r") as f:
    for line in f:
        print(line.strip())
````

## 5. Virtual environments and pip

This part is conceptual — run these in your terminal, not this notebook:

